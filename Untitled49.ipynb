# Data Cleaning and Feature Engineering

```python
madhan['Date'] = pd.to_datetime(
    madhan['Date'],
    dayfirst=True,
    format='mixed',
    errors='coerce'
)

missing_values = madhan.isnull().sum()
print(missing_values)
```

```python
madhan['Discount_Flag'] = np.where(madhan['Discount'] > 0, 1, 0)
madhan['Price_Diff_vs_Competitor'] = madhan['Price'] - madhan['Competitor Pricing']
```

# Exploring Data Analysis(EDA)

```python
madhan.hist(bins=20, figsize=(14,10))
plt.suptitle("Distribution of Key Numerical Variables")
plt.show()
```

```python
sns.boxplot(data=madhan, x='Category', y='Units Sold')
plt.title("Units Sold Distribution by Category")
plt.show()
```

```python
category_sales = madhan.groupby('Category')['Units Sold'].sum().sort_values()

category_sales.plot(kind='barh')
plt.title("Total Units Sold by Category")
plt.xlabel("Units Sold")
plt.show()
```

```python
sns.barplot(data=madhan, x='Region', y='Units Sold', estimator=np.mean)
plt.title("Average Units Sold by Region")
plt.show()
```

```python
sns.boxplot(data=madhan, x='Seasonality', y='Units Sold')
plt.title("Seasonal Impact on Units Sold")
plt.show()
```

```python
sns.barplot(data=madhan, x='Discount_Flag', y='Units Sold')
plt.xticks([0,1], ['No Discount','Discount'])
plt.title("Impact of Discounts on Sales")
plt.show()
```

```python
sns.scatterplot(data=madhan, x='Price', y='Units Sold', hue='Category')
plt.title("Price vs Units Sold")
plt.show()
```

```python
sns.barplot(data=madhan, x='Holiday/Promotion', y='Units Sold')
plt.title("Effect of Promotions on Sales")
plt.show()
```

```python
sns.scatterplot(data=madhan, x='Inventory Level', y='Units Sold')
plt.title("Inventory Level vs Units Sold")
plt.show()
```

```python
madhan['Forecast_Error'] = madhan['Demand Forecast'] - madhan['Units Sold']

sns.histplot(madhan['Forecast_Error'], kde=True)
plt.title("Distribution of Demand Forecast Error")
plt.show()
```

```python
sns.scatterplot(data=madhan, x='Price_Diff_vs_Competitor', y='Units Sold')
plt.title("Price Difference vs Competitor and Sales")
plt.show()
```

```python
sns.boxplot(data=madhan, x='Weather Condition', y='Units Sold')
plt.title("Weather Impact on Sales")
plt.show()
```
